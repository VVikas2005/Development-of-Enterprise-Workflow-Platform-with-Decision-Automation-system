### Prompt 1: Find the second highest salary from the Employee table.

```python
sql_query_1 = """
SELECT MAX(salary) AS SecondHighestSalary
FROM Employee
WHERE salary < (SELECT MAX(salary) FROM Employee);
"""
print(sql_query_1)
```

### Prompt 2: Find all duplicate emails in the Person table.

```python
sql_query_2 = """
SELECT email
FROM Person
GROUP BY email
HAVING COUNT(email) > 1;
"""
print(sql_query_2)
```

### Prompt 3: Find employees who earn more than their managers.

```python
sql_query_3 = """
SELECT e1.name AS Employee
FROM Employee e1
JOIN Employee e2 ON e1.managerId = e2.id
WHERE e1.salary > e2.salary;
"""
print(sql_query_3)
```

### Prompt 4: Find the highest salary in each department.

```python
sql_query_4 = """
SELECT d.name AS Department, e.name AS Employee, e.salary AS Salary
FROM Employee e
JOIN Department d ON e.departmentId = d.id
WHERE (e.departmentId, e.salary) IN (
    SELECT departmentId, MAX(salary)
    FROM Employee
    GROUP BY departmentId
);
"""
print(sql_query_4)
```

### Prompt 5: Find all customers who have never placed an order.

```python
sql_query_5 = """
SELECT name AS Customers
FROM Customers
WHERE id NOT IN (
    SELECT DISTINCT customerId
    FROM Orders
    WHERE customerId IS NOT NULL
);
"""
print(sql_query_5)
```

### Prompt 6: Delete all duplicate email entries in the Person table, keeping only the unique email with the lowest ID.

```python
sql_query_6 = """
DELETE p1 FROM Person p1, Person p2
WHERE p1.email = p2.email AND p1.id > p2.id;
"""
print(sql_query_6)
```

### Prompt 7: Find all employees who earned more than the average salary of their respective department.

```python
sql_query_7 = """
SELECT name, salary, departmentId
FROM Employee e
WHERE salary > (
    SELECT AVG(salary)
    FROM Employee
    WHERE departmentId = e.departmentId
);
"""
print(sql_query_7)
```

### Prompt 8: Find the Nth highest salary from the Employee table.

```python
sql_query_8 = """
SELECT DISTINCT salary
FROM Employee
ORDER BY salary DESC
LIMIT 1 OFFSET 1;
"""
print(sql_query_8)
```

### Prompt 9: List all departments that have fewer than 3 employees.

```python
sql_query_9 = """
SELECT d.name, COUNT(e.id) AS EmployeeCount
FROM Department d
LEFT JOIN Employee e ON d.id = e.departmentId
GROUP BY d.id, d.name
HAVING COUNT(e.id) < 3;
"""
print(sql_query_9)
```

### Prompt 10: Find the consecutive available seats in a cinema (where free is represented by 1).

```python
sql_query_10 = """
SELECT DISTINCT c1.seat_id
FROM cinema c1
JOIN cinema c2 ON ABS(c1.seat_id - c2.seat_id) = 1
WHERE c1.free = 1 AND c2.free = 1
ORDER BY c1.seat_id;
"""
print(sql_query_10)
```

### Prompt 11: Get the cumulative sum of employee salaries ordered by their ID.

```python
sql_query_11 = """
SELECT id, name, salary,
       SUM(salary) OVER(ORDER BY id) AS CumulativeSalary
FROM Employee;
"""
print(sql_query_11)
```

### Prompt 12: Rank employees by salary within each department, allowing gaps if there are ties.

```python
sql_query_12 = """
SELECT departmentId, name, salary,
       RANK() OVER(PARTITION BY departmentId ORDER BY salary DESC) AS SalaryRank
FROM Employee;
"""
print(sql_query_12)
```

### Prompt 13: Find the customer with the highest number of placed orders.

### Prompt 14: Calculate the running average of sales per day over a 3-day window.

```python
sql_query_14 = """
SELECT order_date, daily_amount,
       AVG(daily_amount) OVER(
           ORDER BY order_date
           ROWS BETWEEN 2 PRECEDING AND CURRENT ROW
       ) AS RunningAvg
FROM DailySales;
"""
print(sql_query_14)
```

### Prompt 18: Find the users who logged in on 3 or more consecutive days.

```python
sql_query_18 = """
SELECT DISTINCT l1.user_id
FROM Logins l1
JOIN Logins l2 ON l1.user_id = l2.user_id AND DATEDIFF(l2.login_date, l1.login_date) = 1
JOIN Logins l3 ON l1.user_id = l3.user_id AND DATEDIFF(l3.login_date, l2.login_date) = 1;
"""
print(sql_query_18)
```

### Prompt 19: Find the monthly active users (MAU) for each month.

```python
sql_query_19 = """
SELECT DATE_FORMAT(login_date, '%Y-%m') AS Month,
       COUNT(DISTINCT user_id) AS MonthlyActiveUsers
FROM UserLogins
GROUP BY DATE_FORMAT(login_date, '%Y-%m');
"""
print(sql_query_19)
```

### Prompt 20: Display the first login date for each user.

### Prompt 21: Get the percentage of accepted friend requests for each day.

### Prompt 22: Select employees who have the 3 highest distinct salaries per department.

```python
sql_query_22 = """
SELECT d.name AS Department, e.name AS Employee, e.salary AS Salary
FROM Employee e
JOIN Department d ON e.departmentId = d.id
WHERE (
    SELECT COUNT(DISTINCT e2.salary)
    FROM Employee e2
    WHERE e2.departmentId = e.departmentId AND e2.salary > e.salary
) < 3;
"""
print(sql_query_22)
```

### Prompt 23: Find all dates where the temperature was higher than its previous date's temperature.

```python
sql_query_23 = """
SELECT w1.id
FROM Weather w1
JOIN Weather w2 ON DATEDIFF(w1.recordDate, w2.recordDate) = 1
WHERE w1.temperature > w2.temperature;
"""
print(sql_query_23)
```

### Prompt 24: Pivot the Occupation column in PivotTable so that Name is sorted alphabetically under each Occupation header.

```python
sql_query_24 = """
SELECT
    MAX(CASE WHEN Occupation = 'Doctor' THEN Name END) AS Doctor,
    MAX(CASE WHEN Occupation = 'Professor' THEN Name END) AS Professor,
    MAX(CASE WHEN Occupation = 'Singer' THEN Name END) AS Singer,
    MAX(CASE WHEN Occupation = 'Actor' THEN Name END) AS Actor
FROM (
    SELECT Name, Occupation,
           ROW_NUMBER() OVER(PARTITION BY Occupation ORDER BY Name) AS RowNum
    FROM Occupations
) AS Temp
GROUP BY RowNum;
"""
print(sql_query_24)
```

### Prompt 25: Find the fraction of players that logged in again on the day after they first logged in.

```python
sql_query_25 = """
SELECT ROUND(
    COUNT(a1.player_id) / (SELECT COUNT(DISTINCT player_id) FROM Activity), 2
) AS Fraction
FROM Activity a1
JOIN (
    SELECT player_id, MIN(event_date) AS first_login
    FROM Activity
    GROUP BY player_id
) a2 ON a1.player_id = a2.player_id AND a1.event_date = DATE_ADD(a2.first_login, INTERVAL 1 DAY);
"""
print(sql_query_25)
```
